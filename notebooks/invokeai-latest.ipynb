This notebook allows you to use **Invoke AI** from Google Colab with a complete graphical interface, without local installation or technical knowledge. It is designed so anyone can generate AI images from a browser, regardless of their hardware.

```python
#@markdown ## ⚙️ BLOCK 1 – Prepare the working environment
#@markdown
#@markdown > In this step we'll create the folder where we'll install Invoke and set up the virtual environment.<br>
#@markdown > ⏭️ **Run this cell and wait until it finishes before proceeding to the next one.**

# Create the folder where we will install Invoke AI
!mkdir -p ~/invokeai
%cd ~/invokeai

# Install virtualenv and create the environment with Python 3
!pip install virtualenv
!virtualenv .venv --python=python3
print("----------------------------------------")
print("✅ BLOCK 1 completed successfully. ➡️ Run BLOCK 2 to install Invoke")
```

```python
#@markdown ## 🔧 BLOCK 2 – Install Invoke AI and Cloudflared
#@markdown
#@markdown > We'll install **Invoke AI** and the **Cloudflared** tool to access the web interface from any browser.<br>
#@markdown > This may take a few minutes.<br>
#@markdown > ⏭️ Run this cell and wait for “✅ BLOCK 2 completed successfully.” to appear.<br>
#@markdown > Finally, run the next cell.

# 🔧 BLOCK 2 – Install Invoke AI from PyPI
# Install Invoke AI
.venv/bin/pip install invokeai==6.10.0
print("✅ Invoke 6.10.0 installed successfully.")

# Install Cloudflared
!wget -q https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64.deb
!dpkg -i cloudflared-linux-amd64.deb

print("✅ Cloudflared installed successfully.")
print("----------------------------------------")
print("✅ BLOCK 2 completed successfully. ➡️ Run BLOCK 3 to launch the interface.")
```

```python
#@markdown ## ⚡ Low VRAM Mode (Optional)
#@markdown
#@markdown > Run this cell if you want to enable Low VRAM mode (Recommended)

import os

config_path = "/root/invokeai/invokeai.yaml"

# Create folder if it does not exist
os.makedirs(os.path.dirname(config_path), exist_ok=True)

# Create file with full content
with open(config_path, "w") as f:
    f.write(
"""# Internal metadata - do not edit:
schema_version: 4.0.2

# Put user settings here - see https://invoke-ai.github.io/InvokeAI/configuration/:
enable_partial_loading: true
"""
)

print(f"✔️ File created at {config_path}")
```
